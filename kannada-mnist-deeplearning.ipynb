# Basic Deeplearning explained for MINIST
#### Goal:- To show you how easy it could be to make your first deep leaning model

```python
import pandas as pd
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
from sklearn.model_selection import train_test_split
```

## Loading of data

```python
x = pd.read_csv("/kaggle/input/Kannada-MNIST/train.csv")
y = pd.read_csv("/kaggle/input/Kannada-MNIST/test.csv")
```

## Cleaning and spliting
#### We all know the MNIST dataset is widely recognized for its cleanliness and consistency, making it an ideal benchmark for learners and practitioners.
#### Here we will focus more on building and Evaluating model.


```python
x_train, x_test, y_train, y_test = train_test_split(
    x.drop(columns=["label"]), x["label"])
```

## Training and Fitting the model

```python
# Making a simple model with 
## 1st layer -          with 500 input nodes (input_shape=(784,) -- MNIST images are 28×28, flattened to a 784-length vector. )
## Dropout -            0.2 = 20% of the neurons (Prevents overfitting)
## 2nd hidden layer -   200 input nodes ("relu" is an activation function helps in normalizing the neuron output and give forward)
## 3rd Output layer -   10 output nodes (becase it has 10 numbers to predict {0-9})

model = keras.Sequential([
    keras.layers.Dense(500, input_shape=(784,), activation='relu'),
    keras.layers.Dropout(0.2),
    keras.layers.Dense(200, activation='relu'),
    keras.layers.Dense(10, activation='sigmoid')
])

model.compile(optimizer='adam',  # adjusts learning rates automatically.
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy']) # Tracks the accuracy of predictions
```

```python
model.fit(x_train, y_train, epochs=10)
```

## Evaluation and results

```python
model.evaluate(x_test, y_test)
y_predicted = model.predict(x_test)
# y_predicted
```
